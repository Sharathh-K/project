> **Project focus:** Understanding customer behavior and evaluating the impact of email marketing campaigns using data analysis and causal inference.

## 1. Project Overview

EconoCausal is a data analytics and causal inference project built around the Hillstrom MineThatData E-Mail Analytics dataset.

The objective is to investigate whether email marketing campaigns influence customer behavior, particularly website visits, purchases, and spending.

Instead of simply examining correlations, the project aims to distinguish between customer behavior observed in the campaign groups and the incremental impact that may be attributable to the marketing intervention.

### Key Objectives

- Understand the structure and quality of the raw dataset.
- Prepare the data for reliable analysis.
- Examine differences between treatment and control groups.
- Investigate campaign effects on customer engagement and purchasing behavior.
- Build a reproducible workflow using Python and pandas.

## 3. Understanding the Variables

The dataset contains customer characteristics, historical purchasing information, campaign assignment, and behavioral outcomes.

### A. Customer Characteristics

| Variable | Description |
|---|---|
| `recency` | Time since the customer's last purchase, represented as a recency measure |
| `history_segment` | Category describing the customer's historical spending |
| `history` | Historical customer spending |
| `zip_code` | Customer's geographic classification |

### B. Previous Product Interests

| Variable | Description |
|---|---|
| `mens` | Indicator of previous interest or purchasing history related to men's merchandise |
| `womens` | Indicator of previous interest or purchasing history related to women's merchandise |
| `channel` | Customer's historical purchasing channel |

### C. Treatment and Outcome Variables

| Variable | Role | Meaning |
|---|---|---|
| `segment` | Treatment assignment | Email campaign group or no-email control |
| `visit` | Outcome | Whether the customer visited the website |
| `conversion` | Outcome | Whether the customer made a purchase |
| `spend` | Outcome | Customer spending recorded during the observation period |

**Causal inference note:** `segment` represents the campaign assignment, while `visit`, `conversion`, and `spend` represent outcomes. We must not treat the outcomes as pre-campaign customer characteristics when estimating campaign effects.

### Handling Duplicate Rows

The dataset contains 6,562 exact duplicate rows. These records have not been removed automatically.

Identical rows do not necessarily establish that the same customer was recorded more than once. Without a unique customer identifier, deleting them could remove valid observations.

**Decision:** Retain the original records until the duplicate issue can be investigated further.
