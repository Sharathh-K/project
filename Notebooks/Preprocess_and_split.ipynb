```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('../Data/Raw/Kevin_Hillstrom_MineThatData_E-MailAnalytics_DataMiningChallenge_2008.03.20.csv')
```

```python
df.head()
```

```python
df['zip_code']=df['zip_code'].replace('Surburban', 'Suburban')
```

Fixed typo "Surburban" to "Suburban"

```python
df.head()
```

```python
df["Customer ID"] = ["C-" + str(i) for i in range(1, len(df) + 1)]
```

```python
df
```

Added Customer ID to make columns to identify uniquely

```python
df.drop(columns='history_segment',inplace=True)
```

```python
df
```

Dropping "history_segment" as they are bins of "history" and nothing different

```python
df.to_csv('../Data/Processed/Hillstorm_prepocessed.csv',index=False)
```

Saved preprocessed data in Data/Processed/Hillstorm_prepocessed.csv

```python
df["history_floor"] = (df["history"] == 29.99).astype(int)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

strata = df["segment"] + "_" + df["conversion"].astype(str)

train, test = train_test_split(df, test_size=0.20, random_state=42, stratify=strata)
train = train.reset_index(drop=True)
test = test.reset_index(drop=True)
```

```python
print(train.shape, test.shape)
print(train.groupby("segment")["conversion"].agg(["size", "sum", "mean"]))
print(test.groupby("segment")["conversion"].agg(["size", "sum", "mean"]))
```

```python
train.to_csv("../Data/Processed/train.csv", index=False)
test.to_csv("../Data/Processed/test.csv", index=False)
```
